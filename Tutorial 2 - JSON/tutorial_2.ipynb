## 1. Writing JSON with Python

```python
import json
```

```python
food_ratings = {
    "organic dog food": 2,
    "human food": 10}

json.dumps(food_ratings)
```

```python
str(food_ratings)
```

```python
numbers_present = {
    1: True,
    2: True,
    3: False}

json.dumps(numbers_present)
```

## 2. Variables

```python
dog_id = 1
dog_name = "Frieda"

dog_registry = {
    dog_id: {
        "name": dog_name
    }}

json.dumps(dog_registry)
```

## 3. Converting to JSON

```python
json.dumps(True)
json.dumps(["eating", "sleeping", "barking"])
```

## 4. Keys

```python
available_nums = {
    (1, 2): True,
    3: False}

json.dumps(available_nums, skipkeys=True)
```
