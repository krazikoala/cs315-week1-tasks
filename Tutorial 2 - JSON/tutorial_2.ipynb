## 1. Writing JSON with Python

```python
import json
```

```python
food_ratings = {
    "organic dog food": 2,
    "human food": 10}

json.dumps(food_ratings)
```

```python
str(food_ratings)
```

```python
numbers_present = {
    1: True,
    2: True,
    3: False}

json.dumps(numbers_present)
```

## 2. Variables

```python
dog_id = 1
dog_name = "Frieda"

dog_registry = {
    dog_id: {
        "name": dog_name
    }}

json.dumps(dog_registry)
```

## 3. Converting to JSON

```python
json.dumps(True)
json.dumps(["eating", "sleeping", "barking"])
```

## 4. Keys

```python
available_nums = {
    (1, 2): True,
    3: False}

json.dumps(available_nums, skipkeys=True)
```

```python
toy_conditions = {
    "chew bone": 7,
    "ball": 3,
    "sock": -1
}

json.dumps(toy_conditions, sort_keys=True)
```

```python
dog_data = {
    "name": "Frieda",
    "is_dog": True,
    "hobbies": ["eating", "sleeping", "barking"],
    "age": 8,
    "address": {
        "work": None,
        "home": ("Berlin", "Germany")
    },
    "friends": [
        {
            "name": "Philipp",
            "hobbies": ["eating", "sleeping", "reading"]
        },
        {
            "name": "Mitch",
            "hobbies": ["running", "snacking"]
        }
    ]}
```

```python
with open("hello_frieda.json", mode="w", encoding="utf-8") as write_file:
    json.dump(dog_data, write_file)
```

```python
dog_registry = {
    1: {
        "name": "Frieda"
    }}
dog_json = json.dumps(dog_registry)

dog_json
```

```python
new_dog_registry = json.loads(dog_json)
new_dog_registry
```

## 5. Python to JSON

```python
dog_data = {
    "name": "Frieda",
    "is_dog": True,
    "hobbies": ["eating", "sleeping", "barking"],
    "age": 8,
    "address": {
        "work": None,
        "home": ("Berlin", "Germany")
    }}
```

```python
dog_data_json = json.dumps(dog_data)
dog_data_json
```

```python
new_dog_data = json.loads(dog_data_json)
new_dog_data
```

```python
type(dog_data["address"]["home"])
```

```python
type(new_dog_data["address"]["home"])
```

## 6. Reading JSON File

```python
with open("hello_frieda.json", mode="r", encoding="utf-8") as read_file:
    frie_data = json.load(read_file)
```

```python
type(frie_data)
```

```python
frie_data["name"]
```

```python
frie_data
```

```python
dog_friend = {
    "name": "Mitch",
    "age": 6.5}
```

```python
print(json.dumps(dog_friend))
```

```python
print(json.dumps(dog_friend, indent=2))
```

```python
with open("dog_friend.json", mode="w", encoding="utf-8") as write_file:
    json.dump(dog_friend, write_file, indent=4)
```

```python
with open("dog_friend.json", mode="r", encoding="utf-8") as read_file:
    dog_friend_loaded = json.load(read_file)
dog_friend_loaded
```

```python
with open("hello_frieda.json", mode="r", encoding="utf-8") as input_file:
    original_json = input_file.read()
```

```python
json_data = json.loads(original_json)
```

```python
mini_json = json.dumps(
    json_data,
    indent=None,
    separators=(",", ":"))

mini_json
```

```python
with open("mini_frieda.json", mode="w", encoding="utf-8") as output_file:
    output_file.write(mini_json)
```
